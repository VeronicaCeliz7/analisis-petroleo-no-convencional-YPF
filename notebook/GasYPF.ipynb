```python
col = 'prod_gas'
```

```python
media = df[col].mean()
mediana = df[col].median()



print(f"media:  {media:.2f}")
print(f"mediana: {mediana:.2f}")
```

```python
desviacion = df[col].std()

Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)
IQR = Q3 - Q1
print(f"desviacion estandar: {desviacion:.3f}")
print(f"Q1: {Q1:.2f}")
print(f"Q3: {Q3:.2f}")
print(f"IQR: {IQR:.2f}")

```

```python
ceros = (df[col] == 0).sum() ## estos ceros son registro que no producen gas son ceros
print(f"cantidad de ceros: {ceros}")
positivo = (df[col] > 0 ) .sum()
print(f"cantidad de positivos: {positivo}")
negativo =(df[col] < 0 ) .sum()
print(f"cantidad de negativos: {negativo}")


```

```python
skew = df[col].skew()
kurt = df[col].kurt()


print(f"asimetria: {skew:.2f}")
print(f"curtosis: {kurt:.2f}")
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x=col, color='orange', showfliers=False)
plt.xlabel('Produccion de gas(Miles de m³)')
plt.ylabel('Frecuencia')
plt.title('Distribucion de produccion de gas (Miles de m³)')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x=col, bins=40, color='orange')


plt.xlabel(F'Produccion de gas(Miles m³) {col}')
plt.ylabel('Frecuencia')
plt.title(F'Distribucion de Produccion de Gas (Miles de m³) {col}')
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x=col, bins=40, color='orange')
plt.yscale('log')
plt.xlim(0,5000)
plt.xlabel(f'Produccion de gas(Miles m³) {col}')
plt.ylabel('Frecuencia')
plt.title(F'Distribucion de Produccion de Gas (Miles m³) {col}  en escala logaritmica.')
```
